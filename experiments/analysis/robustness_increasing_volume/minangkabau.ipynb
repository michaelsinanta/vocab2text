```python
from functools import lru_cache
import matplotlib.pyplot as plt
import pandas as pd
import re 

DICTIONARIES_DIR = "../../dictionaries"
NS = [1, 2, 3, 4, 5]


def tokenize(text: str):
    return re.findall(r"\w+", str(text).lower())


@lru_cache(maxsize=None)
def load_dictionary(language):
    dictionary_df = pd.read_csv(f"{DICTIONARIES_DIR}/{language}.csv")
    words = set()
    for word in dictionary_df["word"].dropna().astype(str):
        words |= set(tokenize(word))
    return words


def ngram_stats(texts, dictionary_words, ns=NS):
    unique_valid = {n: set() for n in ns}
    total = {n: 0 for n in ns}

    for t in texts:
        tokens = tokenize(t)
        is_valid = [w in dictionary_words for w in tokens]

        for n in ns:
            for i in range(len(tokens) - n + 1):
                total[n] += 1
                if all(is_valid[i:i + n]):
                    unique_valid[n].add(tuple(tokens[i:i + n]))

    stats = {}
    for n in ns:
        n_unique = len(unique_valid[n])
        stats[f"total_{n}grams"] = total[n]
        stats[f"unique_valid_{n}grams"] = n_unique
        stats[f"distinct_{n}"] = (
            n_unique / total[n] if total[n] > 0 else 0
        )

    stats["avg_distinct_1_5"] = (
        sum(stats[f"distinct_{n}"] for n in NS) / len(NS)
    )

    return stats

checkpoints = list(range(100, 1001, 100))   # 100, 200, ..., 1000
group_cols = ["language", "model", "is_thinking_on", "prompt_type"]

rows = []
for keys, g in df_all_5_words.groupby(group_cols):
    outputs = g["output"].dropna().astype(str).tolist()
    for n in checkpoints:
        if len(outputs) < n:
            break
        dictionary_words = load_dictionary("minangkabau")
        bg = ngram_stats(outputs[:n], dictionary_words)
        rows.append({
            **dict(zip(group_cols, keys)),
            "n_sentences": n,
            **bg,
            "avg_distinct_1_5": bg["avg_distinct_1_5"],
            "pct_unique_ngram": bg["avg_distinct_1_5"] * 100,
        })

stress_df = pd.DataFrame(rows)
display(stress_df)
stress_df.to_csv("stress_test_distinct_ngram_5_words.csv", index=False, encoding="utf-8")

# ---- plot: rows = language, cols = model; color = prompt_type, dashed = thinking on ----
languages = sorted(stress_df["language"].unique())
models = sorted(stress_df["model"].unique())
prompt_types = sorted(stress_df["prompt_type"].unique())

palette = {
    "few_shots_and_word_candidates": {
        False: "#0072B2",
        True: "#56B4E9",
    },
    "few_shots_only": {
        False: "#D55E00",
        True: "#E69F00",
    },
    "topic_related": {
        False: "#009E73",
        True: "#CC79A7",
    },
}

extra = iter(["#000000", "#F0E442", "#999999"])
for pt in prompt_types:
    if pt not in palette:
        palette[pt] = {False: next(extra), True: next(extra)}

fig, axes = plt.subplots(
    len(languages),
    len(models),
    figsize=(5.5 * len(models), 3.4 * len(languages)),
    sharex=True,
    squeeze=False,
)

for i, lang in enumerate(languages):
    for j, model in enumerate(models):
        ax = axes[i][j]
        sub = stress_df[
            (stress_df["language"] == lang) &
            (stress_df["model"] == model)
        ]

        for (pt, think), s in sub.groupby(
            ["prompt_type", "is_thinking_on"]
        ):
            s = s.sort_values("n_sentences")
            ax.plot(
                s["n_sentences"],
                s["pct_unique_ngram"],
                color=palette[pt][bool(think)],
                linestyle="--" if think else "-",
                marker="s" if think else "o",
                markersize=4,
                linewidth=2,
                label=f"{pt} | thinking={'on' if think else 'off'}",
            )

        if i == 0:
            ax.set_title(model)
        if j == 0:
            ax.set_ylabel(f"{lang}\n% average distinct n-gram (n=1–5)")
        if i == len(languages) - 1:
            ax.set_xlabel("# of generated sentences")

        ax.set_xticks(checkpoints)
        ax.tick_params(axis="x", labelrotation=45, labelsize=7)
        ax.grid(alpha=0.3)

handles, labels = axes[0][0].get_legend_handles_labels()
fig.legend(
    handles,
    labels,
    loc="lower center",
    ncol=3,
    fontsize=8,
    frameon=False,
)

fig.suptitle(
    "Stress test: average distinct n-gram (n=1–5) vs. number of generated sentences",
    y=1.0,
)
fig.tight_layout(rect=[0, 0.07, 1, 0.98])
fig.savefig(
    "stress_test_distinct_ngram_5_words_minangkabau.png",
    dpi=200,
    bbox_inches="tight",
)
plt.show()
```
